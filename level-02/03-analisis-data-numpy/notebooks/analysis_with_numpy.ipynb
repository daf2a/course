# Analysis with NumPy

## 1. Mengolah Dataset

```python
import numpy as np

cabang = np.array(["Jakarta", "Bandung", "Surabaya", "Medan"])
bulan = np.array(["Jan", "Feb", "Mar"])

penjualan = np.array([
    [120, 135, 128],
    [98, 108, 112],
    [145, 138, 152],
    [87, 95, 103]
], dtype=np.int64)

target_bulan = np.array([110, 115, 120])
```

### Properti array

Periksa dimensi, ukuran, dan tipe data.

```python
print("Jumlah dimensi:", penjualan.ndim)
print("Bentuk:", penjualan.shape)
print("Jumlah nilai:", penjualan.size)
print("Tipe data:", penjualan.dtype)
```

### Indexing dan slicing

Pilih satu nilai, baris, kolom, atau rentang.

```python
print("Jakarta pada Februari:", penjualan[0, 1])
print("Semua cabang pada Januari:", penjualan[:, 0])
print("Bandung dan Medan pada Februari-Maret:")
print(penjualan[1:4:2, 1:])
```

### `reshape` dan transpose

`reshape` mengubah bentuk. `.T` menukar baris dan kolom.

```python
raw_q1 = np.array([
    120, 135, 128,
    98, 108, 112,
    145, 138, 152,
    87, 95, 103
])

tabel_cabang = raw_q1.reshape(4, 3)
tabel_bulan = tabel_cabang.T

print("Per cabang:", tabel_cabang.shape)
print("Per bulan:", tabel_bulan.shape)
```

## 2. Operasi vektor

Operasi diterapkan pada setiap elemen array.

```python
penjualan_setelah_diskon = penjualan * 0.95
selisih_dengan_target = penjualan - target_bulan

print(penjualan_setelah_diskon)
print(selisih_dengan_target)
```

## 3. Broadcasting per bulan

Satu faktor berlaku untuk kolom bulan yang sama pada semua cabang.

```python
faktor_bulan = np.array([1.00, 1.04, 1.08])
proyeksi_q1 = penjualan * faktor_bulan

print("Bentuk penjualan:", penjualan.shape)
print("Bentuk faktor:", faktor_bulan.shape)
print("Bentuk proyeksi:", proyeksi_q1.shape)
print(proyeksi_q1)
```

### Broadcasting per cabang

Ubah faktor ke bentuk `(4, 1)` agar satu faktor berlaku per baris.

```python
faktor_cabang = np.array([1.05, 0.98, 1.08, 0.97])
faktor_cabang_kolom = faktor_cabang.reshape(4, 1)
proyeksi_per_cabang = penjualan * faktor_cabang_kolom

print("Bentuk faktor:", faktor_cabang_kolom.shape)
print("Bentuk hasil:", proyeksi_per_cabang.shape)
```

## 4. Boolean mask dan `np.where`

Gunakan kondisi untuk memilih nilai atau memberi label.

```python
mask_januari = penjualan[:, 0] >= 100
cabang_januari_kuat = cabang[mask_januari]

mask_target_bulanan = proyeksi_q1 >= target_bulan
nilai_di_atas_target = proyeksi_q1[mask_target_bulanan]

print(cabang_januari_kuat)
print(nilai_di_atas_target)
```

### Label status target

Gunakan `np.where` untuk memberi status tiap nilai.

```python
status_bulanan = np.where(
    proyeksi_q1 >= target_bulan,
    "Mencapai target",
    "Di bawah target"
)

print(status_bulanan)
```

## 5. Agregasi dengan `axis`

Pilih `axis` untuk menghitung total per cabang atau per bulan.

```python
total_per_cabang = proyeksi_q1.sum(axis=1)
total_per_bulan = proyeksi_q1.sum(axis=0)
rata_per_cabang = proyeksi_q1.mean(axis=1)

cabang_tertinggi = cabang[np.argmax(total_per_cabang)]
bulan_tertinggi = bulan[np.argmax(total_per_bulan)]

print("Total per cabang:", total_per_cabang)
print("Total per bulan:", total_per_bulan)
print("Cabang dengan total tertinggi:", cabang_tertinggi)
print("Bulan dengan total tertinggi:", bulan_tertinggi)
```

## 6. Menggabungkan data April

Tambahkan array April sebagai kolom dengan `np.concatenate`.

```python
april_dasar = np.array([132, 117, 160, 108]).reshape(4, 1)
penjualan_4bulan = np.concatenate((penjualan, april_dasar), axis=1)

print(penjualan_4bulan.shape)
print(penjualan_4bulan)
```
