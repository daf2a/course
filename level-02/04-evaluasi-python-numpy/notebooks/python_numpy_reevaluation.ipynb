## 1. Python fundamentals

- **Statement:** satu instruksi yang dijalankan Python, misalnya `total = harga * jumlah`.
- **Variabel:** nama untuk menyimpan nilai. Operator `=` memberi atau memperbarui nilai variabel.
- **Tipe data:** menentukan cara nilai dipakai. Angka dapat dihitung, sedangkan teks dapat digabungkan atau ditampilkan.
- **Input dan output:** `input()` menerima masukan sebagai `str`. `print()` menampilkan nilai dan dapat menerima beberapa nilai yang dipisahkan koma.
- **Casting:** gunakan `int()` untuk bilangan bulat, `float()` untuk bilangan desimal, dan `str()` untuk teks. `int(3.8)` menghasilkan `3`, bukan pembulatan menjadi `4`.
- **Komentar:** diawali `#` dan tidak dijalankan. Gunakan untuk memberi konteks atau menjelaskan langkah penting.
- **Indentation:** menentukan isi suatu blok, seperti isi `if`, loop, atau function. Gunakan indentation yang konsisten.
- **Urutan operasi:** perkalian dan pembagian dikerjakan sebelum penjumlahan dan pengurangan. Tanda kurung dapat memperjelas urutan perhitungan.

**Tipe data yang digunakan**

| Tipe | Contoh | Penggunaan |

| `float` | `0.35` | Berat, jarak, atau hasil rata-rata |
| `str` | `"Jakarta"` | Nama dan teks |
| `bool` | `True`, `False` | Status atau hasil kondisi |
| `list` | `[80, 75, 90]` | Kumpulan nilai berurutan |

**Operator aritmatika**

| Operator | Fungsi | Contoh dan hasil |
| --- | --- | --- |
| `+` | Penjumlahan | `10 + 3` menghasilkan `13` |
| `-` | Pengurangan | `10 - 3` menghasilkan `7` |
| `*` | Perkalian | `10 * 3` menghasilkan `30` |

## 2. Conditional statements

- **Kondisi:** ekspresi yang menghasilkan `True` atau `False`, misalnya `jumlah <= stok`.
- **`if`:** menjalankan blok ketika kondisi benar. **`else`:** menjalankan pilihan alternatif ketika kondisi sebelumnya tidak terpenuhi.
- **`elif`:** memeriksa pilihan berikutnya. Dalam satu rangkaian `if / elif / else`, hanya blok pertama yang cocok yang dijalankan.
- **Urutan pemeriksaan:** letakkan aturan sesuai prioritas. Contohnya, jumlah pesanan tidak valid diperiksa sebelum menghitung subtotal.
- **Beberapa `if` terpisah:** dapat menjalankan lebih dari satu blok. Cocok ketika satu produk memiliki beberapa masalah yang semuanya perlu dicatat.
- **`and`:** benar jika semua kondisi benar. Contoh `produk_aktif and jumlah <= stok` memerlukan produk aktif sekaligus stok yang cukup.
- **`or`:** benar jika minimal satu kondisi benar. Contoh `not kemasan_utuh or not label_lengkap` mendeteksi salah satu masalah atau keduanya.
- **`not`:** membalik nilai Boolean. Contoh `not tersedia` benar ketika `tersedia` bernilai `False`.
- **Nested `if`:** pemeriksaan di dalam pemeriksaan lain. Berguna jika langkah lanjutan hanya berlaku setelah syarat awal terpenuhi.
- **`match/case`:** memilih aksi berdasarkan satu nilai, seperti kode layanan. `case _` menjadi pilihan default.
- **`pass`:** placeholder untuk blok yang belum diisi. Blok tersebut valid, tetapi tidak melakukan tindakan.
- **Batas kondisi:** `>=` dan `<=` menyertakan batas. `480 <= berat <= 520` menerima berat tepat `480` maupun `520`.

**Operator pembanding**

| Operator | Arti | Contoh |
| --- | --- | --- |
| `==` | Sama dengan | `status == "paid"` |
| `!=` | Tidak sama dengan | `status != "batal"` |
| `>` | Lebih besar | `peserta > kapasitas` |
| `<` | Lebih kecil | `stok < stok_minimum` |
| `>=` | Lebih besar atau sama dengan | `nilai >= 80` |
| `<=` | Lebih kecil atau sama dengan | `jumlah <= stok` |

## 3. Loops

- **`for`:** mengolah setiap item dalam urutan, misalnya setiap penjualan pada list. Nilai item berubah pada setiap putaran.
- **`while`:** mengulang selama kondisi benar. Cocok untuk proses yang berhenti karena keadaan berubah, seperti stok yang semakin berkurang.
- **Perubahan keadaan:** perbarui counter, indeks, atau stok yang dipakai pada kondisi `while` agar proses dapat mencapai kondisi berhenti.
- **`range(start, stop, step)`:** membuat urutan angka. `stop` tidak ikut dihasilkan. `range(1, 5)` menghasilkan `1, 2, 3, 4`.
- **Langkah `range()`:** `range(0, 6, 2)` menghasilkan `0, 2, 4`. Langkah negatif dapat digunakan untuk hitung mundur.
- **`enumerate(data)`:** menghasilkan indeks dan nilai item. Indeks dimulai dari `0`, atau dari nilai yang diberikan melalui `start=1` untuk nomor tampilan.
- **Accumulator:** variabel yang diperbarui selama loop. Total biasanya dimulai dari `0`, kemudian ditambah nilai setiap item.
- **Pencarian nilai terbesar:** simpan kandidat terbesar, lalu perbarui jika item berikutnya lebih besar. Untuk data tidak kosong, kandidat awal dapat berasal dari item pertama.
- **Nested loops:** loop luar membaca kelompok atau baris, sedangkan loop dalam membaca item di dalamnya. Contohnya, membaca setiap kursi pada setiap baris bioskop.
- **Posisi dan nilai:** indeks menunjukkan letak item, sedangkan item menunjukkan datanya. Indeks dapat digunakan untuk membaca label pada list lain yang urutannya sesuai.

**Kontrol pengulangan**

| Instruksi | Pengaruh | Contoh penggunaan |
| --- | --- | --- |
| `break` | Keluar dari loop tempat instruksi dijalankan | Menghentikan pelayanan ketika pesanan berikutnya tidak dapat dipenuhi |
| `continue` | Melewati sisa putaran saat ini dan menuju putaran berikutnya | Melewati transaksi batal saat menghitung transaksi selesai |

## 4. Functions and statistics

- **Function:** memberi nama pada langkah yang dapat dipakai berulang. `def` mendefinisikan function dan indentation menentukan isinya.
- **Parameter:** nama input pada definisi function, seperti `data` dan `batas_lulus`. Argument adalah nilai yang diberikan saat function dipanggil.
- **`return`:** mengembalikan hasil agar dapat disimpan, dibandingkan, atau dipakai dalam perhitungan berikutnya. `return` juga mengakhiri pemanggilan function tersebut.
- **`print()`:** menampilkan hasil. Function yang hanya mencetak dan tidak mengembalikan nilai menghasilkan `None` sebagai nilai kembaliannya.
- **Penggunaan ulang:** satu function dapat menerima dataset atau batas yang berbeda. Aturan perhitungan tidak perlu ditulis ulang untuk setiap kelas.
- **Input kosong:** tentukan hasil yang jelas sebelum menghitung. Dalam practice ini, function statistik mengembalikan `None` untuk data kosong.
- **Recursion:** function memanggil dirinya sendiri dengan masalah yang lebih kecil. Base case menentukan kapan pemanggilan berhenti.
- **Fibonacci:** base case `f(0) = 0` dan `f(1) = 1`. Nilai berikutnya mengikuti `f(n) = f(n - 1) + f(n - 2)`, sehingga `f(4) = 3`.
- **Population variance:** seluruh data yang diberikan dianggap sebagai populasi untuk latihan ini, sehingga pembaginya adalah jumlah data `n`.
- **Interpretasi variasi:** dua dataset dapat memiliki mean yang sama tetapi standard deviation berbeda. Standard deviation lebih kecil menunjukkan nilai yang lebih berdekatan terhadap mean.

**Ringkasan statistik**

| Ukuran | Cara menghitung | Makna |
| --- | --- | --- |
| Mean | Jumlah nilai dibagi jumlah data | Rata-rata seluruh nilai |
| Median | Urutkan data dan ambil nilai tengah. Untuk jumlah genap, rata-ratakan dua nilai tengah | Titik tengah data yang sudah diurutkan |
| Mode / modus | Cari nilai dengan frekuensi tertinggi | Nilai yang paling sering muncul. Beberapa data dapat memiliki lebih dari satu modus |
| Range / rentang | Nilai terbesar dikurangi nilai terkecil | Jarak antara kedua ujung data |
| Population variance | Jumlah kuadrat selisih setiap nilai terhadap mean, dibagi `n` | Ukuran variasi dalam satuan yang dikuadratkan |
| Standard deviation | Akar kuadrat variance, menggunakan `math.sqrt()` | Ukuran variasi dalam satuan yang sama dengan data asli |

## 5. Lists and dictionaries

- **List:** kumpulan item berurutan. Posisi item disebut indeks dan dimulai dari `0`.
- **Indexing:** `data[0]` mengambil item pertama. Indeks negatif membaca dari belakang, sehingga `data[-1]` mengambil item terakhir.
- **Slicing:** `data[start:stop:step]` mengambil sebagian urutan. Batas `stop` tidak disertakan. `data[:2]` mengambil dua item pertama.
- **Perubahan posisi:** menghapus item membuat item setelahnya bergeser. Karena itu, posisi harus dibaca berdasarkan isi list saat ini.
- **Nested list:** list yang berisi list lain. `nilai[1][2]` membaca item ketiga pada baris kedua.
- **Dictionary:** menyimpan pasangan key dan value. Contohnya, `produk["stok"]` membaca nilai untuk key `stok`.
- **`.items()`:** membaca key dan value secara berpasangan, sehingga cocok untuk mencetak setiap bagian ringkasan.
- **List of dictionaries:** setiap dictionary mewakili satu record, seperti satu produk atau satu peserta. Loop membaca record, lalu key memilih atributnya.
- **Membentuk hasil baru:** buat list kosong, pilih record yang memenuhi kondisi, lalu tambahkan hasil dengan `append()`.
- **Function bawaan:** `len()` menghitung item, `sum()` menjumlahkan angka, `min()` dan `max()` mencari nilai terkecil serta terbesar, dan `sorted()` menghasilkan list yang diurutkan.

**Operasi list**

| Operasi | Pengaruh | Contoh |
| --- | --- | --- |
| `append(item)` | Menambahkan satu item ke akhir list | `antrean.append("Tono")` |
| `remove(item)` | Menghapus kemunculan pertama berdasarkan nilai | `antrean.remove("Salsa")` |
| `pop(indeks)` | Menghapus dan mengembalikan item pada posisi tertentu | `dilayani = antrean.pop(0)` |
| `pop()` | Menghapus dan mengembalikan item terakhir | `item_terakhir = antrean.pop()` |
| `copy()` | Membuat salinan list luar | `ringkasan = antrean.copy()` |

## 6. NumPy

- **Array:** kumpulan data yang dapat dihitung per elemen. Array 2D dapat mewakili baris stasiun dan kolom hari, atau baris siswa dan kolom komponen nilai.
- **Dimensi dan shape:** array berbentuk `(4, 3)` memiliki empat baris dan tiga kolom. Shape `(2, 4, 3)` dapat mewakili dua periode, empat stasiun, dan tiga hari.
- **Indexing:** `data[0, 1]` mengambil satu nilai pada baris pertama dan kolom kedua.
- **Slicing:** `data[:, 1]` mengambil kolom kedua, sedangkan `data[1:3, :]` mengambil baris berindeks `1` dan `2`. Batas akhir tidak disertakan.
- **`reshape()`:** mengubah bentuk tanpa mengubah jumlah elemen. Array berisi 12 nilai dapat diubah menjadi `(4, 3)`.
- **Transpose `.T`:** untuk array 2D, menukar baris dan kolom. Shape `(4, 3)` berubah menjadi `(3, 4)`.
- **Operasi vektor:** perhitungan diterapkan per elemen. Contohnya, `jumlah_terjual * harga_produk` menghitung pendapatan setiap produk.
- **Broadcasting:** bentuk dibandingkan dari dimensi paling kanan. Ukuran pada tiap dimensi harus sama atau salah satunya `1`. Dimensi yang belum ada diperlakukan berukuran `1` untuk pencocokan.
- **Faktor per kolom:** array `(3,)` dapat dipakai dengan matriks `(4, 3)` karena tiga nilainya mengikuti tiga kolom.
- **Faktor per baris:** ubah empat faktor menjadi `(4, 1)` untuk matriks `(4, 3)`. Bentuk `(4,)` tidak cocok untuk operasi tersebut, karena dimensi terakhirnya berbeda.
- **Boolean mask:** perbandingan menghasilkan array `True` dan `False`. Mask per baris dapat memilih label atau baris yang memenuhi syarat.
- **`np.where(kondisi, hasil_benar, hasil_salah)`:** menghasilkan nilai atau label sesuai kondisi pada setiap posisi.
- **`np.argmax()`:** menghasilkan indeks nilai terbesar. Gunakan indeks itu untuk mengambil label stasiun, cabang, atau siswa.
- **`np.concatenate()`:** menggabungkan array. Untuk array 2D, `axis=0` menambah baris dan `axis=1` menambah kolom. Ukuran pada dimensi lainnya harus cocok.

**Properti array**

| Properti | Fungsi | Contoh untuk array `(4, 3)` |
| --- | --- | --- |
| `ndim` | Jumlah dimensi | `2` |
| `shape` | Ukuran setiap dimensi | `(4, 3)` |
| `size` | Jumlah seluruh elemen | `12` |
| `dtype` | Tipe elemen dalam array | Misalnya `int64` atau `float64` |

**Agregasi pada array 2D berbentuk `(4, 3)`**

| Operasi | Yang dirangkum | Bentuk hasil | Contoh makna |
| --- | --- | --- | --- |
| `data.sum(axis=0)` | Baris, sehingga tersisa satu total untuk setiap kolom | `(3,)` | Total pendapatan per produk |
| `data.sum(axis=1)` | Kolom, sehingga tersisa satu total untuk setiap baris | `(4,)` | Total pendapatan per cabang |
| `data.mean(axis=0)` | Baris, sehingga tersisa satu rata-rata untuk setiap kolom | `(3,)` | Rata-rata suhu per hari |
| `data.mean(axis=1)` | Kolom, sehingga tersisa satu rata-rata untuk setiap baris | `(4,)` | Rata-rata suhu per stasiun |
