# List, Dictionary, dan NumPy

## 1. List

### Indexing, negative indexing, dan slicing

```python
warna = ["biru", "kuning", "hijau", "merah"]

print(warna[0])
print(warna[-1])
print(warna[1:3])
```

### `append()`, `remove()`, dan `pop()`

```python
angka = [10, 20, 30]

angka.append(40)
angka.remove(20)
item_terakhir = angka.pop()

print(angka)
print("Item yang diambil:", item_terakhir)
```

### Function bawaan untuk membaca list

```python
nilai = [72, 85, 90, 68]

print("Jumlah data:", len(nilai))
print("Terendah:", min(nilai))
print("Tertinggi:", max(nilai))
print("Total:", sum(nilai))
print("Urut:", sorted(nilai))
print("Rata-rata:", sum(nilai) / len(nilai))
```

## 2. Nested List

### Mengambil data dari list dua dimensi

```python
nilai = [
    [80, 90, 85],
    [70, 75, 80],
    [90, 95, 88]
]

print(nilai[0])
print(nilai[0][1])
```

### Loop pada setiap baris data

```python
for siswa in nilai:
    rata_rata = sum(siswa) / len(siswa)
    print(rata_rata)
```

## 3. Dictionary

### Key dan value

```python
siswa = {
    "nama": "Alya",
    "umur": 16,
    "nilai": [80, 90, 85]
}

print(siswa["nama"])
print(siswa["nilai"])
```

### Membaca pasangan data dengan `.items()`

```python
for key, value in siswa.items():
    print(key, value)
```

## 4. List of Dictionary

### Dataset sederhana

```python
data_siswa = [
    {"nama": "Alya", "nilai": 85},
    {"nama": "Budi", "nilai": 72},
    {"nama": "Citra", "nilai": 91}
]

print(data_siswa[0]["nama"])
print(data_siswa[2]["nilai"])
```

### Mengolah dataset dengan loop

```python
total = 0
tertinggi = data_siswa[0]

for siswa in data_siswa:
    total += siswa["nilai"]

    if siswa["nilai"] > tertinggi["nilai"]:
        tertinggi = siswa

rata_rata = total / len(data_siswa)
print("Rata-rata:", rata_rata)
print("Nilai tertinggi:", tertinggi["nama"])

print("Nilai minimal 80:")
for siswa in data_siswa:
    if siswa["nilai"] >= 80:
        print(siswa["nama"])
```

## 5. Intro NumPy

### Menghitung rata-rata secara manual

```python
data = [72, 85, 90, 68]

total = 0
for nilai in data:
    total += nilai

mean = total / len(data)
print(mean)
```

### NumPy menyediakan operasi statistik

```python
import numpy as np

data = np.array([72, 85, 90, 68])

print(np.mean(data))
print(np.median(data))
print(np.std(data))
```

### Hasil manual dan NumPy dapat dibandingkan

```python
mean_manual = sum(data) / len(data)
mean_numpy = np.mean(data)

print("Manual:", mean_manual)
print("NumPy:", mean_numpy)
print("Sama:", mean_manual == mean_numpy)
```
