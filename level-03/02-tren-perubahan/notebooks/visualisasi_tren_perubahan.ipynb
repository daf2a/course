# Visualisasi Data: Tren dan Perubahan

Kita akan memilih grafik untuk data yang memiliki urutan waktu, membandingkan tren beberapa kelompok, dan menunjukkan perubahan bertahap.

## Persiapan

Notebook ini memakai pandas, Matplotlib, dan Plotly. Data yang dipakai adalah contoh buatan untuk latihan membaca pola.

```python
import pandas as pd
import matplotlib.pyplot as plt


plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 100)
```

## Membaca data berurutan waktu

Satu baris mewakili satu cabang pada satu bulan. Urutan bulan perlu dibuat eksplisit sebelum menggambar garis.

```python
bulan = ['Jan', 'Feb', 'Mar', 'Apr', 'Mei', 'Jun']
penjualan = pd.DataFrame({
    'Bulan': bulan,
    'Utara': [42, 47, 45, 56, 62, 60],
    'Tengah': [35, 39, 44, 43, 51, 57],
    'Selatan': [28, 33, 31, 40, 44, 49],
})
penjualan
```

### Line chart menampilkan perubahan satu nilai sepanjang waktu

Pakai line chart saat urutan waktu penting dan fokusnya satu seri.

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(penjualan['Bulan'], penjualan['Utara'], marker='o', color='#196B9B')
ax.set(title='Penjualan Cabang Utara', xlabel='Bulan', ylabel='Unit terjual')
plt.show()
```

### Multi-line membandingkan tren beberapa kelompok

Gunakan warna dan legenda untuk membedakan seri. Terlalu banyak garis membuat pola sulit dibaca.

```python
fig, ax = plt.subplots(figsize=(8, 4))
for cabang, warna in zip(['Utara', 'Tengah', 'Selatan'], ['#196B9B', '#D99A37', '#55B77A']):
    ax.plot(penjualan['Bulan'], penjualan[cabang], marker='o', label=cabang, color=warna)
ax.set(title='Tren Penjualan per Cabang', xlabel='Bulan', ylabel='Unit terjual')
ax.legend(title='Cabang')
plt.show()
```

## Area chart menunjukkan besar nilai sepanjang waktu

Area memberi penekanan visual pada besaran. Untuk satu seri, area membantu melihat perubahan total dari waktu ke waktu.

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.fill_between(penjualan['Bulan'], penjualan['Utara'], color='#4FB6E8', alpha=0.45)
ax.plot(penjualan['Bulan'], penjualan['Utara'], marker='o', color='#196B9B')
ax.set(title='Area Penjualan Cabang Utara', xlabel='Bulan', ylabel='Unit terjual')
plt.show()
```

### Stacked area menunjukkan total dan kontribusi seri

Setiap lapisan menambah nilai ke total. Cocok ketika kelompok tidak negatif dan penjumlahan antarkelompok bermakna.

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.stackplot(penjualan['Bulan'], penjualan['Utara'], penjualan['Tengah'], penjualan['Selatan'], labels=['Utara', 'Tengah', 'Selatan'], colors=['#4FB6E8', '#F5D36A', '#55B77A'], alpha=0.9)
ax.set(title='Kontribusi Penjualan per Cabang', xlabel='Bulan', ylabel='Total unit terjual')
ax.legend(loc='upper left', ncol=3)
plt.show()
```

### Stacked area 100% membandingkan proporsi

Setiap bulan dinormalisasi menjadi 100%. Grafik ini memperlihatkan perubahan komposisi, sedangkan perubahan total tidak tampak.

```python
total_bulanan = penjualan[['Utara', 'Tengah', 'Selatan']].sum(axis=1)
persentase_bulanan = penjualan[['Utara', 'Tengah', 'Selatan']].div(total_bulanan, axis=0) * 100

fig, ax = plt.subplots(figsize=(8, 4))
ax.stackplot(
    penjualan['Bulan'],
    persentase_bulanan['Utara'],
    persentase_bulanan['Tengah'],
    persentase_bulanan['Selatan'],
    labels=['Utara', 'Tengah', 'Selatan'],
    colors=['#4FB6E8', '#F5D36A', '#55B77A'],
)
ax.set(title='Komposisi Penjualan per Bulan', xlabel='Bulan', ylabel='Persentase total')
ax.set_ylim(0, 100)
ax.legend(loc='upper left', ncol=3)
plt.show()
```

### Waterfall merangkai kenaikan dan penurunan menuju total

Batang menunjukkan kontribusi positif atau negatif secara bertahap. Grafik ini cocok untuk menjelaskan bagaimana saldo awal berubah menjadi saldo akhir.

```python
perubahan = pd.Series([120, -35, -20, 18, 12], index=['Saldo awal', 'Pembelian', 'Biaya', 'Pengembalian', 'Penyesuaian'])
awal = perubahan.iloc[0]
berjalan = awal
fig, ax = plt.subplots(figsize=(9, 4))
ax.bar(0, awal, color='#196B9B')
for posisi, nilai in enumerate(perubahan.iloc[1:], start=1):
    bawah = berjalan if nilai >= 0 else berjalan + nilai
    ax.bar(posisi, abs(nilai), bottom=bawah, color='#55B77A' if nilai >= 0 else '#D99A37')
    berjalan += nilai
ax.bar(len(perubahan), berjalan, color='#196B9B')
ax.set_xticks(range(len(perubahan) + 1), list(perubahan.index) + ['Saldo akhir'], rotation=20)
ax.set(title='Perubahan Saldo Kas', ylabel='Rupiah (ribu)')
plt.show()
print(f'Saldo akhir: Rp{berjalan:.0f} ribu')
```

### Contoh waterfall saldo acara

Setiap baris adalah perubahan saldo kas, dalam ribuan rupiah. Nilai positif menambah saldo dan nilai negatif menguranginya.

```python
perubahan_acara = pd.Series(
    [-120, -80, 60, -45],
    index=['Sewa tempat', 'Konsumsi', 'Sponsor', 'Perlengkapan'],
)
saldo_awal = 500
saldo_berjalan = saldo_awal
fig, ax = plt.subplots(figsize=(9, 4))
ax.bar(0, saldo_awal, color='#196B9B')
for posisi, nilai_perubahan in enumerate(perubahan_acara, start=1):
    dasar = saldo_berjalan if nilai_perubahan >= 0 else saldo_berjalan + nilai_perubahan
    warna = '#55B77A' if nilai_perubahan >= 0 else '#D99A37'
    ax.bar(posisi, abs(nilai_perubahan), bottom=dasar, color=warna)
    saldo_berjalan += nilai_perubahan
ax.bar(len(perubahan_acara) + 1, saldo_berjalan, color='#196B9B')
label_tahap = ['Saldo awal', *perubahan_acara.index, 'Saldo akhir']
ax.set_xticks(range(len(label_tahap)), label_tahap, rotation=20)
ax.set(title='Perubahan Saldo Kas Acara', ylabel='Rupiah (ribu)')
plt.show()
print(f'Saldo akhir: Rp{saldo_berjalan} ribu')
```
