# Visualisasi Data: Pie Chart dan 100% Stacked Bar

Pie chart merangkum bagian dari satu total. 100% stacked bar membandingkan komposisi per hari setelah setiap total harian diubah menjadi 100%. Contoh dan latihan menggunakan data simulasi Senin sampai Jumat.

*Versi Solusi*

## Demo: waktu membaca beberapa judul novel

Setiap baris merangkum total durasi membaca satu judul pada satu hari. Nilai pada data contoh ini dibuat untuk latihan, bukan catatan aktual.

### Menyiapkan pustaka dan urutan hari

Urutan hari ditetapkan agar tabel dan grafik mengikuti kalender.

```python
import pandas as pd
import matplotlib.pyplot as plt

urutan_hari = ['Senin', 'Selasa', 'Rabu', 'Kamis', 'Jumat']
```

### Data durasi membaca per hari dan judul novel

```python
data_novel = pd.DataFrame(
    [
        ('Senin', 'Laskar Pelangi', 35),
        ('Senin', 'Bumi', 25),
        ('Senin', 'Negeri 5 Menara', 20),
        ('Senin', 'Harry Potter dan Batu Bertuah', 30),
        ('Selasa', 'Laskar Pelangi', 30),
        ('Selasa', 'Bumi', 35),
        ('Selasa', 'Negeri 5 Menara', 25),
        ('Selasa', 'Harry Potter dan Batu Bertuah', 20),
        ('Rabu', 'Laskar Pelangi', 25),
        ('Rabu', 'Bumi', 30),
        ('Rabu', 'Negeri 5 Menara', 15),
        ('Rabu', 'Harry Potter dan Batu Bertuah', 25),
        ('Kamis', 'Laskar Pelangi', 40),
        ('Kamis', 'Bumi', 20),
        ('Kamis', 'Negeri 5 Menara', 30),
        ('Kamis', 'Harry Potter dan Batu Bertuah', 35),
        ('Jumat', 'Laskar Pelangi', 35),
        ('Jumat', 'Bumi', 40),
        ('Jumat', 'Negeri 5 Menara', 20),
        ('Jumat', 'Harry Potter dan Batu Bertuah', 25),
    ],
    columns=['Hari', 'Judul_Novel', 'Durasi_Baca_Menit'],
)

data_novel
```

### Menghitung total durasi per judul

`groupby()` mengumpulkan baris dengan judul yang sama. `sum()` menjumlahkan durasi dari Senin sampai Jumat.

```python
durasi_per_judul = (
    data_novel.groupby('Judul_Novel')['Durasi_Baca_Menit']
    .sum()
    .sort_values(ascending=False)
)

durasi_per_judul
```

### Pie chart untuk melihat bagian dari total mingguan

Setiap irisan menunjukkan bagian durasi baca satu judul dari seluruh durasi baca selama lima hari. Legenda memuat judul novel dan label pada irisan memuat persentasenya.

```python
fig, ax = plt.subplots(figsize=(8, 6))
irisan, _, _ = ax.pie(
    durasi_per_judul.values,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.72,
)
ax.legend(irisan, durasi_per_judul.index, title='Judul novel', loc='center left', bbox_to_anchor=(1, 0.5))
ax.set_title('Komposisi durasi membaca selama lima hari')
ax.axis("equal")
plt.tight_layout()
plt.show()
```

## Practice Tengah: komposisi waktu untuk hobi

Data berikut merangkum durasi setiap hobi per hari. Selesaikan pie chart untuk melihat bagian durasi mingguan yang digunakan pada setiap hobi.

### Data hobi sepulang kegiatan

```python
data_hobi = pd.DataFrame(
    [
        ('Senin', 'Bersepeda', 20),
        ('Senin', 'Berkebun', 10),
        ('Senin', 'Bermain game', 30),
        ('Senin', 'Bermain musik', 15),
        ('Senin', 'Menggambar', 25),
        ('Selasa', 'Bersepeda', 35),
        ('Selasa', 'Berkebun', 15),
        ('Selasa', 'Bermain game', 20),
        ('Selasa', 'Bermain musik', 25),
        ('Selasa', 'Menggambar', 15),
        ('Rabu', 'Bersepeda', 15),
        ('Rabu', 'Berkebun', 15),
        ('Rabu', 'Bermain game', 25),
        ('Rabu', 'Bermain musik', 20),
        ('Rabu', 'Menggambar', 35),
        ('Kamis', 'Bersepeda', 25),
        ('Kamis', 'Berkebun', 10),
        ('Kamis', 'Bermain game', 30),
        ('Kamis', 'Bermain musik', 30),
        ('Kamis', 'Menggambar', 20),
        ('Jumat', 'Bersepeda', 30),
        ('Jumat', 'Berkebun', 20),
        ('Jumat', 'Bermain game', 25),
        ('Jumat', 'Bermain musik', 20),
        ('Jumat', 'Menggambar', 25),
    ],
    columns=['Hari', 'Hobi', 'Durasi_Menit'],
)

data_hobi
```

### Tugas pie chart

1. Jumlahkan `Durasi_Menit` untuk setiap `Hobi`.
2. Buat pie chart dengan legenda nama hobi dan persentase pada irisan.
3. Cetak hobi dengan total durasi tertinggi.

```python
durasi_per_hobi = (
    data_hobi.groupby('Hobi')['Durasi_Menit']
    .sum()
    .sort_values(ascending=False)
)
print("Total durasi per hobi (menit):")
print(durasi_per_hobi.to_string())

fig, ax = plt.subplots(figsize=(8, 6))
irisan, _, _ = ax.pie(
    durasi_per_hobi.values,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.72,
)
ax.legend(irisan, durasi_per_hobi.index, title='Hobi', loc='center left', bbox_to_anchor=(1, 0.5))
ax.set_title('Komposisi durasi hobi selama lima hari')
ax.axis("equal")
plt.tight_layout()
plt.show()

print(f"Hobi dengan durasi mingguan tertinggi: {durasi_per_hobi.idxmax()} ({durasi_per_hobi.max()} menit).")
```

## 100% Stacked Bar membandingkan komposisi harian

Setiap batang mewakili satu hari dan selalu mencapai 100%. Segmen menunjukkan persentase durasi setiap judul terhadap total baca hari itu. Gunakan tabel durasi untuk melihat perbedaan total menit antarhari.

### Menyusun tabel hari dan judul novel

`pivot()` menempatkan hari sebagai baris dan judul novel sebagai kolom. Tabel ini masih menyimpan durasi dalam menit.

```python
tabel_novel_harian = data_novel.pivot(
    index='Hari',
    columns='Judul_Novel',
    values='Durasi_Baca_Menit',
).reindex(
    index=urutan_hari,
    columns=['Laskar Pelangi', 'Bumi', 'Negeri 5 Menara', 'Harry Potter dan Batu Bertuah'],
)

tabel_novel_harian
```

### Mengubah durasi menjadi persentase per hari

Bagi nilai setiap judul dengan total durasi pada hari yang sama, lalu kalikan 100. Jumlah persentase pada setiap baris menjadi 100%.

```python
persentase_novel_harian = tabel_novel_harian.div(
    tabel_novel_harian.sum(axis=1),
    axis=0,
) * 100

print("Persentase durasi baca setiap judul per hari:")
print(persentase_novel_harian.round(1))
print("\nJumlah persentase per hari:")
print(persentase_novel_harian.sum(axis=1).round(1))
```

### Membuat 100% stacked bar

Gunakan tabel persentase sebagai tinggi segmen. Sumbu y menunjukkan porsi durasi dari 0% sampai 100%.

```python
ax = persentase_novel_harian.plot(
    kind='bar',
    stacked=True,
    figsize=(11, 5),
    colormap='tab20',
)
ax.set_xlabel('Hari')
ax.set_ylabel('Porsi durasi baca (%)')
ax.set_title('Komposisi novel dalam waktu baca setiap hari')
ax.set_ylim(0, 100)
ax.set_yticks(range(0, 101, 20))
ax.set_yticklabels([f'{nilai}%' for nilai in range(0, 101, 20)])
ax.legend(title='Judul novel', bbox_to_anchor=(1.02, 1), loc='upper left')
ax.tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()
```

## Practice Akhir: komposisi waktu hobi setiap hari

Gunakan `data_hobi` yang sama dengan Practice Tengah. Ubah durasi harian menjadi persentase, lalu bandingkan porsi setiap hobi dari Senin sampai Jumat.

### Tugas 100% stacked bar dan interpretasi

1. Susun tabel dengan `Hari` sebagai baris, `Hobi` sebagai kolom, dan `Durasi_Menit` sebagai nilai. Urutkan hari Senin sampai Jumat.
2. Bagi setiap nilai dengan total pada hari yang sama, lalu kalikan 100.
3. Tampilkan tabel persentase dan buat 100% stacked bar dengan judul, legend, label sumbu, serta skala 0% sampai 100%.
4. Tampilkan hobi dengan porsi terbesar per hari dan total durasi hobi setiap hari. Jelaskan informasi yang terlihat pada grafik dan tabel.

```python
urutan_hobi = ['Bersepeda', 'Berkebun', 'Bermain game', 'Bermain musik', 'Menggambar']
tabel_hobi_harian = data_hobi.pivot(
    index='Hari',
    columns='Hobi',
    values='Durasi_Menit',
).reindex(
    index=urutan_hari,
    columns=urutan_hobi,
)

persentase_hobi_harian = tabel_hobi_harian.div(
    tabel_hobi_harian.sum(axis=1),
    axis=0,
) * 100

print("Persentase durasi hobi setiap hari:")
print(persentase_hobi_harian.round(1))

ax = persentase_hobi_harian.plot(
    kind='bar',
    stacked=True,
    figsize=(11, 5),
    colormap='tab20',
)
ax.set_xlabel('Hari')
ax.set_ylabel('Porsi durasi hobi (%)')
ax.set_title('Komposisi hobi yang dilakukan setiap hari')
ax.set_ylim(0, 100)
ax.set_yticks(range(0, 101, 20))
ax.set_yticklabels([f'{nilai}%' for nilai in range(0, 101, 20)])
ax.legend(title='Hobi', bbox_to_anchor=(1.02, 1), loc='upper left')
ax.tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()

porsi_terbesar_per_hari = persentase_hobi_harian.idxmax(axis=1)
total_hobi_per_hari = tabel_hobi_harian.sum(axis=1)
print("Hobi dengan porsi terbesar per hari:")
print(porsi_terbesar_per_hari)
print("\nTotal durasi hobi per hari (menit):")
print(total_hobi_per_hari)
```

- 100% stacked bar membandingkan porsi kategori pada beberapa hari.
- Baca total menit harian dari tabel karena tinggi batang 100% stacked selalu sama.
