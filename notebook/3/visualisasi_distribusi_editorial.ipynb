# Visualisasi Data: Distribusi

Kita akan membaca bentuk sebaran nilai, membandingkan kelompok, dan memilih grafik yang menunjukkan frekuensi, kepadatan, serta ringkasan distribusi.

**Versi Solusi**

## Persiapan

Notebook ini memakai pandas, Matplotlib, dan Plotly. Data yang dipakai adalah contoh buatan untuk latihan membaca pola.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde


plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 100)
```

## Mengenali bentuk distribusi

Satu baris mewakili satu peserta dan satu nilai ujian. Data contoh sengaja memuat beberapa nilai rendah dan sebagian besar nilai di rentang menengah sampai tinggi.

```python
nilai = pd.Series([52, 58, 61, 64, 65, 67, 68, 69, 70, 71, 72, 73, 74, 75, 76, 77, 78, 80, 82, 86], name='Nilai')
print(nilai.describe().round(1))
```

### Histogram mengelompokkan nilai ke dalam bins

Tinggi batang menunjukkan frekuensi pada rentang nilai. Lebar dan jumlah bins memengaruhi detail pola yang terlihat.

```python
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(nilai, bins=6, color='#4FB6E8', edgecolor='white')
ax.set(title='Distribusi Nilai Ujian', xlabel='Nilai', ylabel='Jumlah peserta')
plt.show()
```

### KDE memperhalus frekuensi menjadi kurva kepadatan

KDE merangkum area nilai yang relatif padat. Bandwidth mengatur kehalusan kurva dan dapat menyamarkan puncak yang berdekatan.

```python
rentang = np.linspace(nilai.min() - 5, nilai.max() + 5, 250)
kepadatan = gaussian_kde(nilai)(rentang)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(rentang, kepadatan, color='#196B9B', linewidth=2.5)
ax.fill_between(rentang, kepadatan, color='#4FB6E8', alpha=0.25)
ax.set(title='Kepadatan Nilai Ujian', xlabel='Nilai', ylabel='Kepadatan')
plt.show()
```

## Practice Tengah: waktu menyelesaikan kuis

Setiap angka adalah waktu peserta dalam menit. Buat dua histogram dengan jumlah bins berbeda dan satu KDE. Bandingkan informasi yang ditonjolkan histogram dan kurva KDE. Jangan menghapus nilai ekstrem tanpa alasan.

```python
waktu = pd.Series([8, 9, 10, 10, 11, 11, 12, 12, 13, 13, 14, 14, 15, 16, 17, 18, 21, 25], name='Menit')
fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)
for ax, jumlah_bins in zip(axes, [5, 10]):
    ax.hist(waktu, bins=jumlah_bins, color='#4FB6E8', edgecolor='white')
    ax.set(title=f'{jumlah_bins} bins', xlabel='Waktu (menit)', ylabel='Peserta')
plt.tight_layout()
plt.show()

rentang_waktu = np.linspace(waktu.min() - 3, waktu.max() + 3, 250)
kepadatan_waktu = gaussian_kde(waktu)(rentang_waktu)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(rentang_waktu, kepadatan_waktu, color='#196B9B', linewidth=2.5)
ax.fill_between(rentang_waktu, kepadatan_waktu, color='#4FB6E8', alpha=0.25)
ax.set(title='KDE Waktu Kuis', xlabel='Waktu (menit)', ylabel='Kepadatan')
plt.show()
print('Histogram memperlihatkan hitungan per rentang dan bentuknya berubah saat jumlah bins berubah. KDE memberi ringkasan bentuk yang lebih halus. Beberapa peserta menyelesaikan kuis lebih lama dari mayoritas.')
```

## Box plot merangkum median, kuartil, dan pencilan

Box menunjukkan kuartil pertama sampai ketiga, garis di dalamnya adalah median, dan whisker merangkum rentang sesuai aturan plot. Titik di luar whisker ditandai sebagai calon pencilan.

```python
kelas_merah = [62, 66, 68, 70, 71, 73, 74, 76, 81, 92]
kelas_biru = [55, 60, 64, 67, 70, 74, 78, 82, 86, 89]
kelas_hijau = [48, 55, 59, 62, 64, 67, 69, 72, 80, 94]
nilai_per_kelas = [kelas_merah, kelas_biru, kelas_hijau]
nama_kelas = ['Merah', 'Biru', 'Hijau']

fig, ax = plt.subplots(figsize=(8, 4))
plot = ax.boxplot(nilai_per_kelas, tick_labels=nama_kelas, patch_artist=True)
for kotak, warna in zip(plot['boxes'], ['#EAF6FC', '#FFF7D6', '#E8F4EC']):
    kotak.set_facecolor(warna)
ax.set(title='Ringkasan Nilai per Kelas', ylabel='Nilai')
plt.show()
```

### Violin plot menggabungkan ringkasan kelompok dan bentuk kepadatan

Lebar violin menunjukkan kepadatan relatif nilai. Median dan kuartil membantu membaca pusat dan sebaran di dalam bentuk tersebut.

```python
fig, ax = plt.subplots(figsize=(8, 4))
plot = ax.violinplot(nilai_per_kelas, showmedians=True, showextrema=True)
for badan, warna in zip(plot['bodies'], ['#4FB6E8', '#F5D36A', '#55B77A']):
    badan.set_facecolor(warna)
    badan.set_alpha(0.62)
ax.set_xticks([1, 2, 3], nama_kelas)
ax.set(title='Bentuk Distribusi Nilai per Kelas', ylabel='Nilai')
plt.show()
```

### Median serupa dapat menyembunyikan sebaran berbeda

Setiap angka adalah nilai satu peserta dari dua kelompok contoh. Median keduanya sama, tetapi histogram menunjukkan lebar sebaran yang berbeda.

```python
nilai_renggang = np.array([55, 60, 65, 70, 74, 78, 85, 90, 95, 100])
nilai_rapat = np.array([68, 72, 73, 74, 76, 76, 77, 78, 79, 82])
bins_nilai = [50, 60, 70, 80, 90, 100, 110]

fig, axes = plt.subplots(1, 2, figsize=(10, 4), sharex=True, sharey=True)
for ax, data, nama, warna in zip(
    axes,
    [nilai_renggang, nilai_rapat],
    ['Sebaran renggang', 'Nilai terkumpul'],
    ['#4FB6E8', '#F5D36A'],
):
    median = np.median(data)
    ax.hist(data, bins=bins_nilai, color=warna, edgecolor='white')
    ax.axvline(median, color='#196B9B', linestyle='--', label=f'Median {median:.0f}')
    ax.set(title=nama, xlabel='Nilai', ylabel='Jumlah peserta')
    ax.legend()
plt.tight_layout()
plt.show()
```

### Nilai di luar whisker perlu diperiksa

Aturan IQR menandai kandidat pencilan, bukan kesalahan data. Periksa nilai mentah, satuan, dan proses pengumpulan sebelum membuat keputusan.

```python
waktu_penyelesaian = pd.Series([28, 31, 32, 33, 34, 35, 35, 36, 37, 39, 45, 92], name='Menit')
q1 = waktu_penyelesaian.quantile(0.25)
q3 = waktu_penyelesaian.quantile(0.75)
iqr = q3 - q1
batas_bawah = q1 - 1.5 * iqr
batas_atas = q3 + 1.5 * iqr
kandidat = waktu_penyelesaian[(waktu_penyelesaian < batas_bawah) | (waktu_penyelesaian > batas_atas)]

fig, ax = plt.subplots(figsize=(7, 3.5))
ax.boxplot(waktu_penyelesaian, tick_labels=['Waktu kuis'], showfliers=True)
ax.set(title='Waktu Penyelesaian Kuis', ylabel='Menit')
plt.show()
print('Kandidat pencilan menurut aturan IQR:', kandidat.tolist())
```

## Membaca pusat, sebaran, dan bentuk bersama

Median atau rata-rata merangkum pusat, kuartil atau rentang menunjukkan sebaran, dan histogram, KDE, serta violin membantu membaca bentuk. Baca ringkasan tersebut bersama data mentah dan konteksnya.

### Memilih grafik distribusi

Histogram cocok untuk membaca frekuensi. KDE menonjolkan bentuk yang halus. Box plot meringkas pusat dan sebaran beberapa kelompok. Violin plot menambahkan bentuk kepadatan, tetapi perlu dijelaskan agar lebarnya tidak disalahartikan sebagai jumlah data.

## Practice Akhir: nilai tiga kelas

Setiap angka adalah nilai siswa. Buat histogram total, KDE total, box plot per kelas, dan violin plot per kelas. Bandingkan median dan rentang sebaran. Periksa nilai ekstrem sebelum menyimpulkan adanya pencilan.

```python
nilai_kelas = {
    'Merah': [54, 60, 63, 65, 67, 68, 70, 73, 78, 90],
    'Biru': [61, 64, 66, 68, 69, 70, 71, 72, 75, 77],
    'Hijau': [48, 55, 59, 62, 64, 67, 69, 72, 80, 94],
}
semua_nilai = np.concatenate(list(nilai_kelas.values()))
fig, ax = plt.subplots(figsize=(8, 4))
ax.hist(semua_nilai, bins=8, color='#4FB6E8', edgecolor='white')
ax.set(title='Distribusi Nilai Tiga Kelas', xlabel='Nilai', ylabel='Jumlah siswa')
plt.show()

rentang_semua = np.linspace(semua_nilai.min() - 5, semua_nilai.max() + 5, 250)
kepadatan_semua = gaussian_kde(semua_nilai)(rentang_semua)
fig, ax = plt.subplots(figsize=(8, 4))
ax.plot(rentang_semua, kepadatan_semua, color='#196B9B', linewidth=2.5)
ax.fill_between(rentang_semua, kepadatan_semua, color='#4FB6E8', alpha=0.25)
ax.set(title='KDE Nilai Tiga Kelas', xlabel='Nilai', ylabel='Kepadatan')
plt.show()

fig, ax = plt.subplots(figsize=(8, 4))
ax.boxplot(list(nilai_kelas.values()), tick_labels=list(nilai_kelas.keys()), patch_artist=True, boxprops={'facecolor': '#EAF6FC'})
ax.set(title='Box Plot Nilai per Kelas', ylabel='Nilai')
plt.show()

fig, ax = plt.subplots(figsize=(8, 4))
ax.violinplot(list(nilai_kelas.values()), showmedians=True, showextrema=True)
ax.set_xticks([1, 2, 3], list(nilai_kelas.keys()))
ax.set(title='Violin Plot Nilai per Kelas', ylabel='Nilai')
plt.show()
ringkasan = pd.DataFrame({kelas: pd.Series(skor).describe() for kelas, skor in nilai_kelas.items()}).loc[['50%', 'min', 'max']].T
print(ringkasan.rename(columns={'50%': 'Median'}).round(1))
print('Median ketiga kelas berdekatan. Kelas Hijau memiliki rentang terluas, tetapi nilai ekstrem perlu diperiksa dalam konteks sebelum diberi label pencilan.')
```
