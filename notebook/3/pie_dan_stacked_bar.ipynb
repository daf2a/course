# Visualisasi Data: Pie Chart dan 100% Stacked Bar

Pie chart merangkum bagian dari satu total. 100% stacked bar membandingkan komposisi per hari setelah setiap total harian diubah menjadi 100%. Contoh dan latihan menggunakan data simulasi Senin sampai Jumat.

## Demo: waktu membaca beberapa judul novel

Setiap baris merangkum total durasi membaca satu judul pada satu hari. Nilai pada data contoh ini dibuat untuk latihan, bukan catatan aktual.

### Menyiapkan pustaka dan urutan hari

Urutan hari ditetapkan agar tabel dan grafik mengikuti kalender.

```python
import pandas as pd
import matplotlib.pyplot as plt

urutan_hari = ['Senin', 'Selasa', 'Rabu', 'Kamis', 'Jumat']
```

### Data durasi membaca per hari dan judul novel

```python
data_novel = pd.DataFrame(
    [
        ('Senin', 'Laskar Pelangi', 35),
        ('Senin', 'Bumi', 25),
        ('Senin', 'Negeri 5 Menara', 20),
        ('Senin', 'Harry Potter dan Batu Bertuah', 30),
        ('Selasa', 'Laskar Pelangi', 30),
        ('Selasa', 'Bumi', 35),
        ('Selasa', 'Negeri 5 Menara', 25),
        ('Selasa', 'Harry Potter dan Batu Bertuah', 20),
        ('Rabu', 'Laskar Pelangi', 25),
        ('Rabu', 'Bumi', 30),
        ('Rabu', 'Negeri 5 Menara', 15),
        ('Rabu', 'Harry Potter dan Batu Bertuah', 25),
        ('Kamis', 'Laskar Pelangi', 40),
        ('Kamis', 'Bumi', 20),
        ('Kamis', 'Negeri 5 Menara', 30),
        ('Kamis', 'Harry Potter dan Batu Bertuah', 35),
        ('Jumat', 'Laskar Pelangi', 35),
        ('Jumat', 'Bumi', 40),
        ('Jumat', 'Negeri 5 Menara', 20),
        ('Jumat', 'Harry Potter dan Batu Bertuah', 25),
    ],
    columns=['Hari', 'Judul_Novel', 'Durasi_Baca_Menit'],
)

data_novel
```

### Menghitung total durasi per judul

`groupby()` mengumpulkan baris dengan judul yang sama. `sum()` menjumlahkan durasi dari Senin sampai Jumat.

```python
durasi_per_judul = (
    data_novel.groupby('Judul_Novel')['Durasi_Baca_Menit']
    .sum()
    .sort_values(ascending=False)
)

durasi_per_judul
```

### Pie chart untuk melihat bagian dari total mingguan

Setiap irisan menunjukkan bagian durasi baca satu judul dari seluruh durasi baca selama lima hari. Legenda memuat judul novel dan label pada irisan memuat persentasenya.

```python
fig, ax = plt.subplots(figsize=(8, 6))
irisan, _, _ = ax.pie(
    durasi_per_judul.values,
    autopct="%1.1f%%",
    startangle=90,
    pctdistance=0.72,
)
ax.legend(irisan, durasi_per_judul.index, title='Judul novel', loc='center left', bbox_to_anchor=(1, 0.5))
ax.set_title('Komposisi durasi membaca selama lima hari')
ax.axis("equal")
plt.tight_layout()
plt.show()
```

## 100% Stacked Bar membandingkan komposisi harian

Setiap batang mewakili satu hari dan selalu mencapai 100%. Segmen menunjukkan persentase durasi setiap judul terhadap total baca hari itu. Gunakan tabel durasi untuk melihat perbedaan total menit antarhari.

### Menyusun tabel hari dan judul novel

`pivot()` menempatkan hari sebagai baris dan judul novel sebagai kolom. Tabel ini masih menyimpan durasi dalam menit.

```python
tabel_novel_harian = data_novel.pivot(
    index='Hari',
    columns='Judul_Novel',
    values='Durasi_Baca_Menit',
).reindex(
    index=urutan_hari,
    columns=['Laskar Pelangi', 'Bumi', 'Negeri 5 Menara', 'Harry Potter dan Batu Bertuah'],
)

tabel_novel_harian
```

### Mengubah durasi menjadi persentase per hari

Bagi nilai setiap judul dengan total durasi pada hari yang sama, lalu kalikan 100. Jumlah persentase pada setiap baris menjadi 100%.

```python
persentase_novel_harian = tabel_novel_harian.div(
    tabel_novel_harian.sum(axis=1),
    axis=0,
) * 100

print("Persentase durasi baca setiap judul per hari:")
print(persentase_novel_harian.round(1))
print("\nJumlah persentase per hari:")
print(persentase_novel_harian.sum(axis=1).round(1))
```

### Membuat 100% stacked bar

Gunakan tabel persentase sebagai tinggi segmen. Sumbu y menunjukkan porsi durasi dari 0% sampai 100%.

```python
ax = persentase_novel_harian.plot(
    kind='bar',
    stacked=True,
    figsize=(11, 5),
    colormap='tab20',
)
ax.set_xlabel('Hari')
ax.set_ylabel('Porsi durasi baca (%)')
ax.set_title('Komposisi novel dalam waktu baca setiap hari')
ax.set_ylim(0, 100)
ax.set_yticks(range(0, 101, 20))
ax.set_yticklabels([f'{nilai}%' for nilai in range(0, 101, 20)])
ax.legend(title='Judul novel', bbox_to_anchor=(1.02, 1), loc='upper left')
ax.tick_params(axis='x', rotation=0)
plt.tight_layout()
plt.show()
```
