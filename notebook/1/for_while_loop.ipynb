# For Loop dan While Loop

### 1. `for` dengan list sederhana

`for` mengambil satu item dari list pada setiap putaran.

````python
warna = ["biru", "kuning", "hijau"]

for item in warna:
    print("Warna:", item)
````

### 2. `for` dengan `range()`

`range(1, 4)` menghasilkan angka 1 sampai 3 untuk menentukan jumlah putaran.

````python
for nomor in range(1, 4):
    print("Putaran ke-", nomor)
````

### 3. `while` dengan counter

`while` terus berjalan selama kondisi benar. Counter perlu diperbarui agar loop dapat berhenti.

````python
langkah = 1

while langkah <= 3:
    print("Langkah", langkah)
    langkah = langkah + 1
````

### 4. Nomor urut dengan `enumerate()`

`enumerate()` memberi nomor urut sekaligus nilai dari sebuah list.

````python
playlist = ["Langit Pagi", "Kota Hujan", "Jalan Pulang"]

for urutan, lagu in enumerate(playlist, start=1):
    print(f"{urutan}. {lagu}")
````

### 5. Rekap data dari list

Loop dapat menghitung total dan mencari nilai tertinggi dalam satu kali pembacaan list.

````python
skor_turnamen = [72, 88, 91, 65, 84]
total_skor = 0
skor_tertinggi = skor_turnamen[0]

for skor in skor_turnamen:
    total_skor = total_skor + skor
    if skor > skor_tertinggi:
        skor_tertinggi = skor

rata_rata = total_skor / len(skor_turnamen)
print("Rata-rata:", rata_rata)
print("Tertinggi:", skor_tertinggi)
````

### 6. Filter item dari list

Simpan hanya nilai yang memenuhi kondisi ke list baru.

````python
nilai = [54, 78, 91, 62, 88, 45]
nama = ["Budi", "Sisi", "Dede", "Dimas", "Dika", "Denis"]
lulus = []

for skor in nilai:
    if skor >= 75:
        lulus.append(nama[nilai.index(skor)])

print(lulus)
````

### 7. `range()` dengan langkah dan hitung mundur

Parameter ketiga pada `range()` menentukan besar langkah perubahan angka.

````python
for detik in range(10, 0, -2):
    print(f"Mulai dalam {detik} detik")

print("Mulai!")
````

### 8. Nested loop untuk membuat pola

Loop luar membuat baris. Loop dalam menambahkan simbol pada setiap baris.

````python
tinggi = 5

for baris in range(1, tinggi + 1):
    pola = ""
    for kolom in range(baris):
        pola = pola + "*"
    print(pola)
````

### 9. Nested loop untuk semua kombinasi

Setiap makanan dipasangkan dengan setiap minuman.

````python
makanan = ["Roti", "Donat"]
minuman = ["Teh", "Susu", "Kopi"]

for menu in makanan:
    for minum in minuman:
        print(f"Paket: {menu} + {minum}")
````

### 10. `while` untuk proses berbasis kondisi

Selama baterai masih ada, robot terus mengantar paket.

````python
count = 10

while count > 0:
    print(f"Hitung mundur: {count}")
    count = count - 1
````

### 11. `break` dan `continue` saat mencari data

`continue` melewati data yang tidak dipakai. `break` menghentikan pencarian ketika target ditemukan.

````python
kartu = ["bunga", "hewan", "rusak", "usang", "langka", "roket"]

for item in kartu:
    if item == "usang" or item == "rusak":
        continue
    if item == "langka":
        print("Kartu langka ditemukan!")
        break
    print("Simpan kartu:", item)
````
