## 3. Input dan Output

- `print()` menampilkan output di layar. `input()` menerima teks yang diketik pengguna
- hasil dari `input()` selalu bertipe `str`.

#### `print()`: output dasar

```python
print('Selamat belajar Python!')
print(10 * 2)
```

#### `print()` dengan `,`: memisahkan nilai otomatis

```python
nama = 'Alya'
umur = 17
print('Nama:', nama, '| Umur:', umur)
```

#### `print()` dengan `+`: menggabungkan teks

```python
depan = 'Selamat'
belakang = 'pagi!'
print(depan + ' ' + belakang)
```

#### `print()` dengan `end`: mengatur akhir output

```python
print('Ini jadi')
print('dua baris')

print('Ini jadi', end=' ')
print('satu baris')
```

#### `input()`: menerima teks dari pengguna

```python
nama_pengguna = input('Masukkan nama kamu: ')
print('Halo, ' + nama_pengguna)
```

#### `input()` ke `int`: menerima angka

```python
umur = int(input('Masukkan umur kamu: '))
print('Tahun depan umur kamu', umur + 1)
```

## 4. Casting

Casting mengubah tipe data dengan `int()`, `float()`, atau `str()`.

#### `str` ke `int`: teks bilangan bulat menjadi angka

```python
angka_teks = '25'
angka = int(angka_teks)
print(angka)
print(type(angka))
```

#### `str` ke `float`: teks desimal menjadi angka

```python
jarak_teks = '3.5'
jarak = float(jarak_teks)
print(jarak)
print(type(jarak))
```

#### `int` ke `str`: angka menjadi teks

```python
nomor_antrian = 8
nomor_teks = str(nomor_antrian)
print('Antrian nomor ' + nomor_teks)
print(type(nomor_teks))
```

#### `float` ke `int`: desimal menjadi bilangan bulat

```python
diskon = 9.8
diskon_bulat = int(diskon)
print(diskon_bulat)  # int() menghapus bagian desimal
print(type(diskon_bulat))
```

## 5. Operations

Operator aritmatika digunakan untuk menghitung nilai.

| Operator | Nama | Contoh | Hasil |
| --- | --- | --- | --- |
| `+` | Penjumlahan | `10 + 3` | `13` |
| `-` | Pengurangan | `10 - 3` | `7` |
| `*` | Perkalian | `10 * 3` | `30` |
| `/` | Pembagian | `10 / 3` | `3.333...` |
| `%` | Sisa bagi | `10 % 3` | `1` |
| `**` | Pangkat | `10 ** 2` | `100` |
| `//` | Pembagian bulat | `10 // 3` | `3` |

#### `+`: penjumlahan

```python
angka_pertama = 10
angka_kedua = 3
print('Hasil:', angka_pertama + angka_kedua)
```

#### `-`: pengurangan

```python
stok_awal = 10
barang_terjual = 3
print('Sisa stok:', stok_awal - barang_terjual)
```

#### `*`: perkalian

```python
harga_satuan = 15000
jumlah_pesanan = 2
print('Total harga:', harga_satuan * jumlah_pesanan)
```

#### `/`: pembagian 

```python
total_harga = 45000
jumlah_teman = 3
bayar_per_orang = total_harga / jumlah_teman
print('Bayar per orang:', bayar_per_orang)
```

#### `%`: sisa bagi

```python
jumlah_permen = 10
jumlah_anak = 3
print('Sisa permen:', jumlah_permen % jumlah_anak)
```

#### `**`: pangkat

```python
sisi = 10
print('Luas persegi:', sisi ** 2)
```

#### `//`: pembagian bulat

```python
jumlah_kursi = 10
kursi_per_meja = 3
print('Meja penuh:', jumlah_kursi // kursi_per_meja)
```

## 6. `if`

Kode di dalam `if` hanya berjalan saat kondisi bernilai `True`.

#### `if`: satu kondisi

```python
suhu = 31
if suhu > 30:
    print('Cuaca panas')
```

#### `if`: kondisi pembayaran

```python
saldo = 50000
if saldo >= 25000:
    print('Pembayaran dapat diproses')
```

## 7. `if ... else`

`else` berjalan ketika kondisi `if` salah.

#### `if ... else`: dua kemungkinan

```python
stok = 0
if stok > 0:
    print('Barang tersedia')
else:
    print('Barang habis')
```

#### `if ... else`: validasi kata sandi

```python
kata_sandi = 'python123'
if kata_sandi == 'python123':
    print('Login berhasil')
else:
    print('Kata sandi salah')
```

## 8. `elif`

`elif` mengecek kondisi berikutnya ketika `if` sebelumnya salah.

#### `if / elif / else`: kategori nilai

```python
nilai = 83
if nilai >= 90:
    print('A')
elif nilai >= 75:
    print('B')
else:
    print('C')
```

#### `if / elif / else`: waktu

```python
jam = 14
if jam < 12:
    print('Selamat pagi')
elif jam < 18:
    print('Selamat sore')
else:
    print('Selamat malam')
```

## 9. `pass`

`pass` adalah placeholder: Python tidak melakukan apa pun, tetapi blok kode tetap valid.

#### `pass`: placeholder dalam `if`

```python
fitur_baru = True
if fitur_baru:
    pass

print('Program tetap berjalan')
```

#### `pass`: placeholder dalam `else`

```python
status = 'menunggu'
if status == 'selesai':
    print('Pesanan selesai')
else:
    pass

print('Cek status lagi nanti')
```

## 10. `match/case`

Gunakan `match/case` untuk memilih aksi berdasarkan satu nilai. `case _` adalah pilihan default.

#### `match/case`: memilih hari

```python
hari = 3
match hari:
    case 1:
        print('Senin')
    case 2:
        print('Selasa')
    case 3:
        print('Rabu')
```

#### `match/case`: kode status

```python
kode_status = 404
match kode_status:
    case 200:
        print('OK')
    case 404:
        print('Halaman tidak ditemukan')
    case _:
        print('Status lain')
```

#### `match/case`: beberapa pilihan dengan `|`

```python
hari = 6
match hari:
    case 1 | 2 | 3 | 4 | 5:
        print('Hari kerja')
    case 6 | 7:
        print('Akhir pekan')
    case _:
        print('Nomor hari tidak valid')
```
