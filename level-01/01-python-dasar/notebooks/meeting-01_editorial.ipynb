## Pemeriksaan awal

Jalankan cell berikut untuk menyiapkan feedback otomatis.


````python
def check_equal(label, actual, expected):
    if actual == expected:
        print(f"OK: {label}")
    else:
        print(f"Belum tepat: {label}")
        print(f"  Diperoleh: {actual!r}")
        print(f"  Diharapkan: {expected!r}")


def check_call(label, function, args, expected):
    if not callable(function):
        print(f"Belum tepat: {label}. Function belum dibuat")
        return
    try:
        actual = function(*args)
    except Exception as error:
        print(f"Belum tepat: {label}. Pemanggilan menghasilkan {type(error).__name__}")
        return
    check_equal(label, actual, expected)


print("Pemeriksaan awal selesai. Python siap digunakan.")
````
