## Variable, expression, dan formatted output


```python
item_name = "Buku tulis"
unit_price = 8500
quantity = 4
subtotal = unit_price * quantity
print(f"{quantity} {item_name}: Rp{subtotal}")
```

## Data type dasar

Fokuskan cell ini pada empat data type dasar. 


```python
item_count = 4
item_price = 7500.0
item_name = "Pensil"
is_available = True
print(type(item_count))
print(type(item_price))
print(type(item_name))
print(type(is_available))
```

## Operasi aritmetika

Setelah data type, bandingkan pembagian biasa, pembagian bulat, sisa bagi, pangkat, dan prioritas operator.


```python
left = 17
right = 5
print(left + right)
print(left - right)
print(left * right)
print(left / right)
print(left // right)
print(left % right)
print(2 ** 3)
print(4 + 3 * 2)
print((4 + 3) * 2)
```

## Type conversion


```python
quantity_text = "6"
quantity = int(quantity_text)
price = float("7500.50")
print(quantity_text * 2)
print(quantity * 2)
print(f"Harga: Rp{price:.2f}")
```
