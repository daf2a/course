## Pemeriksaan awal

Jalankan cell berikut untuk menyiapkan feedback otomatis.


````python
def check_equal(label, actual, expected):
    if actual == expected:
        print(f"OK: {label}")
    else:
        print(f"Belum tepat: {label}")
        print(f"  Diperoleh: {actual!r}")
        print(f"  Diharapkan: {expected!r}")


def check_call(label, function, args, expected):
    if not callable(function):
        print(f"Belum tepat: {label}. Function belum dibuat")
        return
    try:
        actual = function(*args)
    except Exception as error:
        print(f"Belum tepat: {label}. Pemanggilan menghasilkan {type(error).__name__}")
        return
    check_equal(label, actual, expected)


print("Pemeriksaan awal selesai. Python siap digunakan.")
````

## `for`, `range()`, dan `while`


````python
for index in range(1, 7, 2):
    print("for", index)

index = 0
while index < 3:
    print("while", index)
    index = index + 1
````

## `break` dan `continue`


````python
for value in [3, -1, 5, 0, 8]:
    if value < 0:
        continue
    if value == 0:
        break
    print(value)
````

## Guided example: trace accumulator


````python
values = [4, 7, 2]
total = 0
for value in values:
    before = total
    total = total + value
    print(value, before, total)
````

## Guided example: minimum dan maksimum


````python
scores = [78, 92, 75, 60, 88]
lowest = scores[0]
highest = scores[0]
for score in scores[1:]:
    if score < lowest:
        lowest = score
    if score > highest:
        highest = score
print(lowest, highest)
````

## Guided example: nested loop


````python
rows = ["A", "B"]
columns = [1, 2, 3]
for row in rows:
    for column in columns:
        print(row, column)
````
