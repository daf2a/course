## Pemeriksaan awal

Jalankan cell berikut untuk menyiapkan feedback otomatis.


````python
def check_equal(label, actual, expected):
    if actual == expected:
        print(f"OK: {label}")
    else:
        print(f"Belum tepat: {label}")
        print(f"  Diperoleh: {actual!r}")
        print(f"  Diharapkan: {expected!r}")


def check_call(label, function, args, expected):
    if not callable(function):
        print(f"Belum tepat: {label}. Function belum dibuat")
        return
    try:
        actual = function(*args)
    except Exception as error:
        print(f"Belum tepat: {label}. Pemanggilan menghasilkan {type(error).__name__}")
        return
    check_equal(label, actual, expected)


print("Pemeriksaan awal selesai. Python siap digunakan.")
````

## Memanggil function dari function lain


````python
def calculate_subtotal(price, quantity):
    return price * quantity

def create_payment_summary(name, price, quantity):
    subtotal = calculate_subtotal(price, quantity)
    return f"{name} membayar Rp{subtotal}"

print(create_payment_summary("Nadia", 15000, 2))
````

## Eksperimen dictionary


````python
product = {"name": "Notebook", "price": 18000, "stock": 5}
print(product["name"])
product["stock"] = 8
product["category"] = "Alat tulis"
print(product.get("publisher", "Tidak ada"))
for key, value in product.items():
    print(key, value)
````

## Guided example: function menerima dan mengembalikan dictionary


````python
def create_product(name, price, stock=0):
    return {"name": name, "price": price, "stock": stock}

def create_product_summary(product):
    return f"{product['name']} | Rp{product['price']} | stock {product['stock']}"

product = create_product("Pensil", 4000, 12)
print(create_product_summary(product))
````

## Guided example: list of dictionaries


````python
products = [
    create_product("Notebook", 18000, 5),
    create_product("Pensil", 4000, 12),
    create_product("Penghapus", 3500, 0)
]
for product in products:
    print(create_product_summary(product))
````
