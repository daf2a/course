# Conditional Statements in Python

## 1. Comparison operators

Comparison operator membandingkan dua nilai dan menghasilkan `True` atau `False`.

| Operator | Arti | Contoh |
| --- | --- | --- |
| `==` | Sama dengan | `nilai == 80` |
| `!=` | Tidak sama dengan | `status != 'selesai'` |
| `>` | Lebih besar | `suhu > 30` |
| `<` | Lebih kecil | `umur < 18` |
| `>=` | Lebih besar atau sama dengan | `nilai >= 75` |
| `<=` | Lebih kecil atau sama dengan | `stok <= 5` |

Ingat: `=` digunakan untuk memberi nilai ke variable, sedangkan `==` digunakan untuk membandingkan nilai.

#### Comparison operators menghasilkan Boolean

```python
angka_pertama = 10
angka_kedua = 7

print('Sama dengan:', angka_pertama == angka_kedua)
print('Tidak sama:', angka_pertama != angka_kedua)
print('Lebih besar:', angka_pertama > angka_kedua)
print('Lebih kecil:', angka_pertama < angka_kedua)
print('Lebih besar atau sama:', angka_pertama >= angka_kedua)
print('Lebih kecil atau sama:', angka_pertama <= angka_kedua)
```

#### Chained comparison: memeriksa sebuah rentang

Python dapat menulis kondisi rentang secara ringkas: `20 <= suhu <= 29`.

```python
suhu = 25

if 20 <= suhu <= 29:
    print('Suhu nyaman')
```

## 2. `if`

Kode di dalam `if` hanya berjalan saat kondisi bernilai `True`.

#### `if`: satu kondisi

```python
suhu = 31
if suhu > 30:
    print('Cuaca panas')
```

#### `if`: kondisi pembayaran

```python
saldo = 50000
if saldo >= 25000:
    print('Pembayaran dapat diproses')
```

## 3. Logical operators: `and`, `or`, dan `not`

Logical operator digunakan untuk menggabungkan atau membalik kondisi.

| Operator | Hasil `True` ketika... |
| --- | --- |
| `and` | Semua kondisi benar |
| `or` | Minimal satu kondisi benar |
| `not` | Kondisi asal bernilai `False` |

```python
umur = 20
punya_tiket = True
hari = 'Sabtu'
sedang_libur = False

print('and:', umur >= 18 and punya_tiket)
print('or:', hari == 'Sabtu' or hari == 'Minggu')
print('not:', not sedang_libur)
```

#### Kondisi gabungan di dalam `if`

```python
nilai = 82
kehadiran = 90

if nilai >= 75 and kehadiran >= 80:
    print("Lulus")

suhu = 31
if suhu < 20 or suhu >= 30:
    print("Suhu berada di luar rentang nyaman")
```

## 4. `if ... else`

`else` berjalan ketika kondisi `if` salah.

#### `if ... else`: dua kemungkinan

```python
stok = 0
if stok > 0:
    print('Barang tersedia')
else:
    print('Barang habis')
```

#### `if ... else`: validasi kata sandi

```python
kata_sandi = 'python123'
if kata_sandi == 'python123':
    print('Login berhasil')
else:
    print('Kata sandi salah')
```

## 5. `elif`

`elif` mengecek kondisi berikutnya ketika `if` sebelumnya salah.

#### `if / elif / else`: kategori nilai

```python
nilai = 83
if nilai >= 90:
    print("A")
elif nilai >= 75:
    print("B")
else:
    print("C")
```

#### `if / elif / else`: waktu

```python
jam = 14
if jam < 12:
    print('Selamat pagi')
elif jam < 18:
    print('Selamat sore')
else:
    print('Selamat malam')
```

## 6. Nested `if`

Nested `if` adalah kondisi di dalam kondisi. Gunakan ketika kondisi kedua hanya perlu diperiksa setelah kondisi pertama benar.

#### Nested `if`: pemeriksaan usia dan tiket

```python
umur = 20
punya_tiket = True

if umur >= 18:
    if punya_tiket:
        print('Boleh masuk')
    else:
        print('Tiket diperlukan')
else:
    print('Usia belum cukup')
```

> Gunakan nested `if` seperlunya. Terlalu banyak tingkat indentasi membuat kode lebih sulit dibaca.

## 8. `pass`

`pass` adalah placeholder: Python tidak melakukan apa pun, tetapi blok kode tetap valid.

#### `pass`: placeholder dalam `if`

```python
fitur_baru = True
if fitur_baru:
    pass

print('Program tetap berjalan')
```

#### `pass`: placeholder dalam `else`

```python
status = 'menunggu'
if status == 'selesai':
    print('Pesanan selesai')
else:
    pass

print('Cek status lagi nanti')
```

## 9. `match/case`

Gunakan `match/case` untuk memilih aksi berdasarkan satu nilai. `case _` adalah pilihan default.

#### `match/case`: memilih hari

```python
hari = 3
match hari:
    case 1:
        print('Senin')
    case 2:
        print('Selasa')
    case 3:
        print('Rabu')
```

#### `match/case`: kode status

```python
kode_status = 404
match kode_status:
    case 200:
        print('OK')
    case 404:
        print('Halaman tidak ditemukan')
    case _:
        print('Status lain')
```

#### `match/case`: beberapa pilihan dengan `|`

```python
hari = 6
match hari:
    case 1 | 2 | 3 | 4 | 5:
        print('Hari kerja')
    case 6 | 7:
        print('Akhir pekan')
    case _:
        print('Nomor hari tidak valid')
```
